```python
merge_1 = pd.merge(df_frequency,df_incidence_gender, on = 'Country', how = 'inner' )
```

```python
merge_1.head()
```

```python
columns_to_drop = ['Male Including NMSC Number','Male Excluding NMSC Number','Female Including NMSC Number', 'Female Excluding NMSC Number']
```

```python
merge_1 = merge_1.drop(columns = columns_to_drop )
```

```python
print(merge_1.columns.to_list())
```

```python
print(df_mortality_gender.columns.to_list())
```

```python
df_mortality_gender = df_mortality_gender.rename(columns = {'Male Including NMSC Rate': 'Male Including NMSC Mortality Rate','Male Excluding NMSC Rate': 'Male Excluding NMSC Mortality Rate', 'Female Including NMSC Rate': 'Female Including NMSC  Rate', 'Female Excluding NMSC Rate': 'Female Excluding NMSC Mortality Rate'  })
```

```python
print(df_mortality_gender.columns.to_list())
```

```python
drop_columns = ['Male Including NMSC Number','Male Excluding NMSC Number', 'Female Including NMSC Number', 'Female Excluding NMSC Number']
```

```python
df_mortality_gender = df_mortality_gender.drop(columns = drop_columns)
```

```python
merge_2 = pd.merge(merge_1,df_mortality_gender, on = 'Country', how = 'inner')
merge_2.head()
```

```python
merge_2 = merge_2.rename(columns = {'Male Including NMSC Rate': 'Male Including NMSC incidence Rate', ' Male Excluding NMSC Rate' : 'Male Excluding NMSC incidence Rate', 'Female Including NMSC Rate': 'Female Including NMSC incidence Rate', 'Female Including NMSC Rate	': ' Female Including NMSC Mortality Rate', 'Female Excluding NMSC Rate': 'Female Excluding NMSC incidence Rate' })
```

```python
print(merge_2.columns.to_list())
```

```python
merge_2.head()
```

```python
sns.barplot(x= 'Country', y= 'Cancer rate', data=merge_2.sort_values('Cancer rate', ascending = False))
plt.xticks(rotation=90)
plt.title('Overall Cancer Rate by Country')
plt.ylabel('Cancer Rate')
plt.show()
```

```python
df = merge_2[['Country', 'Male Including NMSC incidence Rate', 'Female Including NMSC incidence Rate']]
df_melt = df.melt(id_vars = 'Country', var_name = 'Gender', value_name = 'Incidence Rate')

sns.barplot(x = 'Country', y = 'Incidence Rate', hue = 'Gender', data= df_melt)
plt.xticks(rotation=90)
plt.title('Cancer Incidence (Including NMSC) by Gender')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14, 8))
sns.scatterplot(
    x='Male Including NMSC Mortality Rate',
    y='Male Including NMSC incidence Rate',
    hue='Country',
    data=merge_2
)

plt.title('Male Cancer Incidence vs Mortality')
plt.xlabel('Mortality Rate')
plt.ylabel('Incidence Rate')

# Horizontal legend below the plot
plt.legend(
    bbox_to_anchor=(0.5, -0.25),  # Centered below the plot
    loc='upper center',
    ncol=6,  # Adjust number of columns to fit your needs
    frameon=False
)

plt.tight_layout()
plt.show()
```

```python
merge_2['Male Incidence Diff'] = merge_2['Male Including NMSC incidence Rate'] - merge_2['Male Excluding NMSC incidence Rate']
merge_2['Female Incidence Diff'] = merge_2['Female Including NMSC incidence Rate'] - merge_2['Female Excluding NMSC incidence Rate']

# Barplot for difference
sns.barplot(x='Country', y='Male Incidence Diff', data=merge_2)
plt.xticks(rotation=90)
plt.title('Impact of NMSC on Male Incidence Rate')
plt.show()

#Barplot for female difference
sns.barplot(x='Country', y='Female Incidence Diff', data=merge_2)
plt.xticks(rotation=90)
plt.title('Impact of NMSC on Female Incidence Rate')
plt.show()
```

```python
# Sort by Cancer rate
top_countries = merge_2.sort_values('Cancer rate', ascending=False)

plt.figure(figsize=(10,6))
sns.barplot(x='Country', y='Cancer rate', data=top_countries)
plt.xticks(rotation=90)
plt.title('Overall Cancer Rate by Country')
plt.ylabel('Incidence Rate per 100,000')
plt.show()
```
